## 2. Is the dataset good? Does it have enough data?
Explore its size and look for any repeated passenger IDs.


```python
print("Rows and columns:", df.shape)
print("Repeated PassengerId values:", df["PassengerId"].duplicated().sum())
print("Survived values (0 = No, 1 = Yes):")
print(df["Survived"].value_counts())
```

## 3. How many features are there? Is that enough?

## 6 and 8. Which columns and rows have missing (null) values?

```python
missing = df.isnull().sum()
print("Missing values in each column:")
print(missing)
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing > 0).sum())
```

## 7. Which are the features and which is the target?

```python
target = "Survived"
features = df.drop(columns=["Survived", "PassengerId"])

print("Target:", target)
print("Example feature columns:", features.columns.tolist())
print("Target values:", df[target].unique())
```

Answer: The target is Survived (0 = did not survive; 1 = survived). The features are passenger details used to investigate survival. PassengerId is excluded from our example features because it identifies a passenger rather than describing their circumstances. No prediction model is trained in this exercise.
