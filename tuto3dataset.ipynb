```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/StudentsPerformance.csv')
display(df.head())
```

What you should see: Each row represents one student. Each column contains one variable (such as gender, parental level of education, or math score). This is a tidy table in the basic sense: one student per row and one variable per column.


2. Is the dataset good? Does it have enough data?

```python
print("Rows and columns:", df.shape)
print("Repeated student entries (duplicates):", df.duplicated().sum())
print("Test preparation course distribution:")
print(df["test preparation course"].value_counts())
```

Answer: Yes, the dataset is good quality. It contains 1,000 rows and 8 columns with 0 duplicated rows. The data contains a healthy distribution across categories (both students who completed and did not complete test preparation), providing sufficient data to calculate statistics and visualize trends.

3. How many features are there? Is that enough?

```python
print("All columns:")
print(df.columns.tolist())
print("Number of columns:", len(df.columns))
print("Possible input columns (excluding math score):", len(df.columns) - 1)
```

Answer: There are 8 columns: one target outcome (math score) and 7 other feature columns (gender, race/ethnicity, parental level of education, lunch, test preparation course, reading score, writing score). This is enough features to explore how socioeconomic, demographic, and academic factors correlate with student test results.

4. How many observations are there? Is that enough?

```python
print("Number of students:", len(df))
```

Answer: 1,000 observations. That is enough to calculate reliable summary statistics and construct beginner data science visualizations, but whether it is sufficient for high-accuracy production machine learning models depends on the task.

5. What data types are in the dataset?

```python
print(df.dtypes)
```

Explain: int64 represents whole numbers (e.g., test scores like math score, reading score, and writing score). object represents categorical text data (e.g., gender, race/ethnicity, lunch, and test preparation course).

6 and 8. Which columns and rows have missing (null) values?

```python
missing = df.isnull().sum()
print("Missing values in each column:")
print(missing)
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing > 0).sum())
```
