2. Is the dataset good? Does it have enough data?

1 (continued). Make two simple graphs

```python
import matplotlib.pyplot as plt

df["test preparation course"].value_counts().plot(kind="bar", color="cornflowerblue", edgecolor="black")
plt.title("Number of Students by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Number of students")
plt.xticks(rotation=0)
plt.show()
```

```python
df["math score"].plot(kind="hist", bins=10, color="orange", edgecolor="black")
plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of students")
plt.show()
```

Discuss: The number of students who completed the test preparation course is not equal to those who did not (far more students did not complete a prep course). In the math score histogram, the most common score range is around 60 to 75. Unlike datasets with missing records, all 1,000 student scores are represented in the histogram.

9. What useful question can we ask? Does this dataset fit?

Suggested question: How did average math scores differ by test preparation course completion?   This is suitable because test preparation course (none vs completed) and math score are both present. The following table shows averages rather than only student counts.

```python
prep_math_avg = df.groupby("test preparation course")["math score"].mean()
print("Average math score by test preparation course:")
print(prep_math_avg.round(1))

prep_math_avg.plot(kind="bar", color="cornflowerblue", edgecolor="black")
plt.title("Average math score by test preparation course")
plt.xlabel("Test preparation course")
plt.ylabel("Average math score")
plt.xticks(rotation=0)
plt.show()
```

Discussion: Students who completed the preparation course achieved a higher average score (69.7) compared to those who did not take any course (64.1). The dataset fits this question directly because both variables are fully populated without missing entries.
